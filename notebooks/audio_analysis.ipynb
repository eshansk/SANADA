## 1. Load Audio and Plot Waveform

```python
import librosa
import librosa.display
import matplotlib.pyplot as plt

audio_path = "Sample Dataset/recordings/0_george_0.wav"

# Load audio without changing its original sample rate
y, sr = librosa.load(audio_path, sr=None)

print("Sample rate:", sr)
print("Number of samples:", len(y))
print("Duration:", len(y) / sr, "seconds")

# Plot waveform
plt.figure(figsize=(12, 4))
librosa.display.waveshow(y, sr=sr)

plt.title("Audio Waveform")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")

plt.show()
```

## 2. Peak Amplitude, RMS and dBFS

```python
import numpy as np

peak = np.max(np.abs(y))

rms = np.sqrt(np.mean(y**2))

dbfs = 20 * np.log10(rms)

print("Peak amplitude:", peak)
print("RMS:", rms)
print("RMS in dBFS:", dbfs)
```

## 3. RMS Energy Over Time

```python
rms_frames = librosa.feature.rms(y=y)[0]

plt.figure(figsize=(12, 4))
plt.plot(rms_frames)

plt.title("RMS Energy Over Time")
plt.xlabel("Frame")
plt.ylabel("RMS")

plt.show()
```

## 4. Resample Audio (8000 Hz to 16000 Hz)

```python
# Resample audio from 8000 Hz to 16000 Hz
y_16k = librosa.resample(y, orig_sr=sr, target_sr=16000)

print("Original sample rate:", sr)
print("New sample rate: 16000")
print("Original samples:", len(y))
print("New samples:", len(y_16k))
```
