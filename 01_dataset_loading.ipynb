# SANADA: Load Data & Visualize Waveform & Spectrogram
This notebook verifies that the dataset loads properly and visualizes the audio as both a **Waveform** and a **Log-Mel Spectrogram**.

### 1. Import Libraries

```python
import pandas as pd
import soundfile as sf
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np
import IPython.display as ipd

print("Libraries loaded successfully!")
```

### 2. Load and Check 

```python
# Path to the metadata manifest
csv_path = "SANADA/Org_Fsd_Mix/Org_Mix_4_Sounds_Variable/train/metadata.csv"
df = pd.read_csv(csv_path)

print(f"Total mixture files found: {len(df)}")
print("First row details:")
print(df.iloc[0][["mixture_path", "num_speakers", "speaker_labels", "duration_s", "sample_rate_hz"]])
```

### 3. Load Audio File

```python
# Get the audio path from the first row
audio_path = df.iloc[0]["mixture_path"]

# Load audio
waveform, sr = sf.read(audio_path)

print(f"File: {audio_path}")
print(f"Sample Rate: {sr} Hz")
print(f"Total Samples: {len(waveform)} (Duration: {len(waveform)/sr:.1f} seconds)")

# Play audio
ipd.Audio(waveform, rate=sr)
```

### 4. Visualize Waveform and Spectrogram

```python
# 1. Compute Log-Mel Spectrogram
mel_spec = librosa.feature.melspectrogram(y=waveform, sr=sr, n_fft=1024, hop_length=512, n_mels=80)
log_mel = librosa.power_to_db(mel_spec, ref=np.max)

# 2. Plot Waveform and Spectrogram side-by-side
fig, axes = plt.subplots(2, 1, figsize=(10, 6))

# Plot Waveform
librosa.display.waveshow(waveform, sr=sr, ax=axes[0], color="dodgerblue")
axes[0].set_title(f"Waveform: {audio_path} (Speakers: {df.iloc[0]['num_speakers']})")
axes[0].set_ylabel("Amplitude")

# Plot Spectrogram
img = librosa.display.specshow(log_mel, sr=sr, hop_length=512, x_axis="time", y_axis="mel", ax=axes[1], cmap="magma")
fig.colorbar(img, ax=axes[1], format="%+2.0f dB")
axes[1].set_title("Log-Mel Spectrogram (Frequency vs. Time)")

plt.tight_layout()
plt.show()
```
